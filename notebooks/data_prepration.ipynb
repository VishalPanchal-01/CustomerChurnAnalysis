```python

import pandas as pd
import numpy as np
```

```python
file_path = "../data/processed/Telco_Customer_Churn_Dataset.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully.")
```

```python
df.head()
```

```python
print(f"Dataset Shape: {df.shape}")

print(f"Column Names: {df.columns.to_list()}")

print(f"Dataset Information: {df.info()}")
```

```python
print(f"Numerical Statistics: \n{df.describe()}")
print()
print(f"Categorical Statistics: \n{df.describe(include="object")}")
```

```python
# Missing value:
df.isnull().sum()
```

```python
# duplicates rows:
df.duplicated().sum()
```

```python
# check duplicate customers:
duplicate_customers = df["customerID"].duplicated().sum()
duplicate_customers
```

```python
print("\nTotalCharges datatype:")
print(df["TotalCharges"].dtype)

blank_total_charges = (df["TotalCharges"].astype(str).str.strip().eq("").sum())

print("\nBlank TotalCharges values:")
print(blank_total_charges)
```

```python
blank_rows = df[df["TotalCharges"].astype(str).str.strip().eq("")]

print("\nRows containing blank TotalCharges:")
print(blank_rows[["customerID","tenure","MonthlyCharges","TotalCharges","Churn"]])
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"],errors="coerce")
```

```python
print("\nMissing values after TotalCharges conversion:")
print(df.isnull().sum())
```

```python
df["TotalCharges"] = df["TotalCharges"].fillna(0)
```

```python
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

```python
print("\nChurn Distribution:")
print(df["Churn"].value_counts())

print("\nChurn Percentage:")
print(
    df["Churn"]
    .value_counts(normalize=True)
    .mul(100)
)
```

```python
df["Churn"] = df["Churn"].map({"No": 0,"Yes": 1})

print(f"Encoded Churn Distribution:")
print(df["Churn"].value_counts())
```

```python
df_model = df.drop(columns=["customerID"])
```

```python
X = df_model.drop(columns=["Churn"])

y = df_model["Churn"]

print(f"Feature Dataset Shape: \n {X.shape}")

print(f"Target Dataset Shape: \n {y.shape}")
```

```python
numerical_features = X.select_dtypes(include=["int64", "float64"]).columns.tolist()

categorical_features = X.select_dtypes(include=["object"]).columns.tolist()

print(f"Numerical Features: \n {numerical_features}")

print(f"Categorical Features: \n {categorical_features}")
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=42,stratify=y)

print(f"Training Feature Shape: \n {X_train.shape}")

print(f"Testing Feature Shape: \n {X_test.shape}")

print(f"Training Target Shape: \n {y_train.shape}")

print(f"Testing Target Shape: \n {y_test.shape}")

```

```python

print(f"Training Churn Distribution: \n {y_train.value_counts(normalize=True).mul(100)}")
print()
print(f"Testing Churn Distribution: \n {y_test.value_counts(normalize=True).mul(100)}")
```

```python
# Numerical preprocessing pipeline:
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

numeric_transformer = Pipeline(steps=[
        ("imputer",SimpleImputer(strategy="median"))
    ])
```

```python
# categorical preprocessing pipeline:
from sklearn.preprocessing import OneHotEncoder

categorical_transformer = Pipeline(steps=[
        ("imputer",SimpleImputer(strategy="most_frequent")),
        ("onehot",OneHotEncoder(handle_unknown="ignore",drop="first"))
    ])
```

```python
#  Column Transformer :

from sklearn.compose import ColumnTransformer
preprocessor = ColumnTransformer(transformers=[
        ("numerical",numeric_transformer,numerical_features),
        ("categorical",categorical_transformer,categorical_features)
    ])
```

```python
X_train_processed = preprocessor.fit_transform(X_train)

X_test_processed = preprocessor.transform(X_test)


print(f"Processed Training Shape: \n{X_train_processed.shape}")

print(f"Processed Testing Shape: \n{X_test_processed.shape}")
```

```python
feature_names = preprocessor.get_feature_names_out()

print(f"Number of Final Features: \n{len(feature_names)}")

print(f"Processed Feature Names: \n{feature_names}")
```

```python
if hasattr(X_train_processed, "toarray"):
    X_train_processed = X_train_processed.toarray()

if hasattr(X_test_processed, "toarray"):
    X_test_processed = X_test_processed.toarray()


X_train_processed_df = pd.DataFrame(X_train_processed,columns=feature_names,index=X_train.index)

X_test_processed_df = pd.DataFrame(X_test_processed,columns=feature_names,index=X_test.index)
```

```python
print("\nProcessed Training Data:")
print(X_train_processed_df.head())
```

```python
df.to_csv("../data/processed/cleaned_customer_churn.csv",index=False)
```

```python
X_train_processed_df.to_csv("../data/processed/X_train.csv",index=False)

X_test_processed_df.to_csv("../data/processed/X_test.csv",index=False)

y_train.to_csv("../data/processed/y_train.csv",index=False)

y_test.to_csv("../data/processed/y_test.csv",index=False)

```
